## Independence

Sticking with the question of which vectors contain new information, we call $v_1,\ldots,v_k$ **linearly independent** when

$$c_1v_1+\cdots+c_kv_k=0$$

is possible only with $c_1=\cdots=c_k=0$. That all-zero choice is called the trivial combination. A nontrivial zero combination is evidence that one vector can be expressed using the others.

The geometric picture depends on how many vectors we have. Two nonzero vectors pointing along the same line are dependent. Two independent vectors in $\mathbb R^3$ still span only a plane; lying in a plane does not by itself make them dependent. We will use the zero-combination test on concrete matrices next.

Let's consider the example of the matrix (of coefficients), named `A` below.

```python
A= Matrix([[1, 2, 4], [3, 1, 4]])
A
```

### How many basis vectors are needed for the nullspace?

This matrix has $m=2$ rows and $n=3$ columns. The row reduction below shows two pivots, so its rank is $r=2$ (full row rank). That leaves $n-r=3-2=1$ free variable.

This means **one vector in a basis of the nullspace**, not one vector in the whole nullspace. The free variable can take any real value, giving infinitely many solutions to $Ax=0$. Let us see how the software output and the equations describe the same family.

```python
A.rref()  # Reduced row-echelon form
```

```python
A.nullspace()  # A list of basis vectors for the nullspace
```

We note the two equations in three unknowns (set to the zero vector) in (1) below.

$$\begin{align}&{x_1}+2{x_2}+4{x_3}=0\\&3{x_1}+{x_2}+4{x_3}=0\end{align}\tag{1}$$

After Gauss-Jordan elimination we have (2).

$$\begin{align}&{x_1}+\frac{4}{5}{x_3}=0\\&{x_2}+\frac{8}{5}{x_3}=0\end{align}\tag{2}$$

The reduced equations tell us that choosing $x_3$ determines both other coordinates. If we set $x_3=1$, then $x_2=-8/5$ and $x_1=-4/5$. This gives the vector returned in the basis list from `.nullspace()` above.

But the choice 1 was just convenient. With $x_3=t$, the whole solution is

$$x=t\begin{pmatrix}-4/5\\-8/5\\1\end{pmatrix},\qquad t\in\mathbb R.$$

One vector generates the family by scalar multiplication. That is why we say the nullspace has dimension 1, even though it has infinitely many elements.

### Another way to state independence

Consider the columns of any matrix $A$ as vectors $\mathbf{v}_1,\mathbf{v}_2,\ldots,\mathbf{v}_n$.  If $\text{rank}\left(A\right)=n$ (the number of columns) then the nullspace only contains the zero vector and the column vectors are linearly independent.

## Spanning

We have introduced the idea of **spanning** a (sub)space: all linear combinations of a set of vectors fill that space. The set need not be independent. For instance, adding $e_1+e_2$ to the set $\{e_1,e_2\}$ does not change the plane it spans, because that new vector was already available as a combination.

We are particularly interested in a set that fills the space **without redundancy**. It must both span the space and be linearly independent. This leads to the next idea, a **basis**.

## Basis

Basis vectors (in a space $W$) are vectors with the properties, (a) they are linearly independent and (b) they span the space (linear combinations of them fill the space).

Up until now we looked at columns in a matrix $A$.  It is more common in textbooks to look at a space first and ask about basis vectors, spanning vectors, dimension, and so on.

So let's look at $\mathbb{R}^3$.  The obvious set of basis vectors are shown in (3).

$$\hat{i},\quad\hat{j},\quad\hat{k}\tag{3}$$

What about the vectors in (4).  Are they linearly independent and do they span $\mathbb{R}^3$?

$$\begin{bmatrix}1\\1\\2\end{bmatrix},\quad\begin{bmatrix}2\\2\\5\end{bmatrix}\tag{4}$$

```python
A = Matrix([[1, 2], [1, 2], [2, 5]])
A
```

```python
A.rref()
```

Here the two columns are independent: the rank is 2 and there are $n=2$ columns. So $n-r=0$, and the only solution of $Ax=0$ is the zero vector in $\mathbb R^2$. The next `.nullspace()` call returns an empty **basis list**; it does not say that the kernel itself is empty.

Now return to the original question about $\mathbb R^3$. Two independent columns still span only a plane there. They form a basis for that plane, but cannot form a basis for all of $\mathbb R^3$. Independence and spanning the chosen target space are separate questions.

```python
A.nullspace()
```

All their linear combinations will only fill a plane through the origin.  Their (trivial) zero combination does result in the zero vector, though, so they do fill a subspace of $\mathbb{R}^3$.

If we added a column vector that is a linear combination of the original two columns, it will also fall in the plane.  There will be a vector in the nullspace other than the zero vector, though.

```python
A = Matrix([[1, 2, 3], [1, 2, 3], [2, 5, 7]])  # Adding a linear combination of columns one and two
A.nullspace()  # Calculation the nullspace
```

In this square example a zero row after elimination implies fewer pivots than columns, so there is a free variable. For a rectangular matrix, count pivots and columns explicitly; a zero row alone does not imply a free variable.

Let's add another, just for fun.  Here we duplicate the first row.

```python
A = Matrix([[1, 2, 3], [1, 2, 3], [2, 5, 8]])
A
```

```python
A.rref()
```

Again, a column without a pivot and sure enough, we'll find a vector (other than the zero vector) in the nullspace.

```python
A.nullspace()
```

### The special case of a square matrix

If we have a square matrix, we need only look at it's determinant: _Is it invertible_?  (More about matrix inverses later).

```python
A.det() # .det() calculates the determinant
```

The determinant is zero and the kernel contains a **nonzero** vector. Every matrix has the zero vector in its kernel; singularity of a square matrix requires a nonzero one.

## Dimension

Given a (sub)space, every basis for that (sub)space has the same number of vectors (there are usually more than one basis for every (sub)space.  This called the _dimension_ of the (sub)space.

## Example

* Consider the column space

```python
A = Matrix([[1, 2, 3, 1], [1, 1, 2, 1], [1, 2, 3, 1]])
A
```

There are $n=4$ unknowns and $m=3$ equations. The columns live in $\mathbb R^3$; the kernel lives in $\mathbb R^4$. Column 1 equals column 4, so the columns are dependent. Distinguish the ambient spaces before discussing a basis.

```python
A.nullspace()
```

```python
A.rref()
```

As we can see here, (5), columns three and four have free variables, i.e. no pivots.

$$\begin{align}&{x}_{1}+0{x}_{2}+{x}_{3}+{x}_{4}=0\\&0{x}_{1}+1{x}_{2}+{x}_{3}+0{x}_{4}=0\\&{x}_{4}={c}_{2}\\&{x}_{3}={c}_{1}\\ \therefore \quad &{x}_{2}=-{c}_{1}\\ \therefore \quad &{x}_{1}=-{c}_{1}-{c}_{2}\end{align}\tag{5}$$

We express this more fully in (6).

$$  \begin{bmatrix} { x }_{ 1 } \\ { x }_{ 2 } \\ { x }_{ 3 } \\ { x }_{ 4 } \end{bmatrix}=\begin{bmatrix} -{ c }_{ 1 }-{ c }_{ 2 } \\ -{ c }_{ 1 } \\ { c }_{ 1 } \\ { c }_{ 2 } \end{bmatrix}=\begin{bmatrix} -{ c }_{ 1 } \\ -{ c }_{ 1 } \\ { c }_{ 1 } \\ 0 \end{bmatrix}+\begin{bmatrix} -{ c }_{ 2 } \\ 0 \\ 0 \\ { c }_{ 2 } \end{bmatrix}={ c }_{ 1 }\begin{bmatrix} -1 \\ -1 \\ 1 \\ 0 \end{bmatrix}+{ c }_{ 2 }\begin{bmatrix} -1 \\ 0 \\ 0 \\ 1 \end{bmatrix} \tag {6} $$

The two pivots tell us that the rank is 2. To obtain a basis for the **column space**, return to columns 1 and 2 of the original matrix:

$$\begin{pmatrix}1\\1\\1\end{pmatrix},\qquad\begin{pmatrix}2\\1\\2\end{pmatrix}.$$

These vectors live in $\mathbb R^3$ and span the outputs that $A$ can produce. The two vectors calculated above from the homogeneous equations instead form a basis for the **kernel**, which lives in $\mathbb R^4$. The two spaces happen to have the same dimension in this example, but they are not the same space.

In summary, any two independent vectors **in this column space** would form another basis for it. Every such basis has two vectors, so its dimension is the rank $r=2$.

For the kernel we count free variables instead: its dimension is $n-r=4-2=2$. Again, this counts vectors in a basis, not vectors in the whole space. When reading a basis calculation, always ask both questions: “a basis of which space?” and “where do these vectors live?”
