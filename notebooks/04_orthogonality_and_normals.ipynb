## Worked route: choose the coefficient, then explain why

The [original geometry notebook](../2_Geometric_view_of_linear_algebra_checkpoint.ipynb) asks how vectors combine to reach a target. Here the target may lie off the line we can reach, so we ask for the closest reachable point.

Every point on a line through a nonzero vector $u$ is $\alpha u$. If the residual $r=b-\alpha u$ is perpendicular to the line, then

$$u^T(b-\alpha u)=0,\qquad \alpha=\frac{u^Tb}{u^Tu}.$$

For the initial 45-degree line we may use $u=(1,1)^T$ and $b=(3,1)^T$. The coefficient is $4/2=2$, giving projection $(2,2)^T$ and residual $(1,-1)^T$. Their perpendicularity can be checked by a dot product. Using a unit vector in the same direction gives the same projected point.

Why is this the closest point? For any other $tu$, write $b-tu=r+(\alpha-t)u$. The two terms are perpendicular, so their squared lengths add:

$$\|b-tu\|^2=\|r\|^2+(\alpha-t)^2\|u\|^2.$$

The second term cannot be negative, and vanishes exactly at $t=\alpha$. This proves the claim for the whole line, rather than just the plotted points. The seminar sheet changes the direction to $(1,2)$ so you can try the argument again.

```python
def explore(angle=45, bx=3, by=1):
    fig, p, r = la.projection_figure(angle,bx,by)
    print('projection:',p,'residual:',r,'distance:',np.linalg.norm(r))
    display(fig); plt.close(fig)
sliders = la.interactive_plot(explore, {'angle': (-180,180,15,45), 'bx': (-4,4,.5,3), 'by': (-4,4,.5,1)})
```

### From two edges to a normal

For the edges $u=(1,1,0)$ and $v=(0,1,1)$, the component formula gives

$$u\times v=(1\cdot1-0\cdot1,\;0\cdot0-1\cdot1,\;1\cdot1-1\cdot0)=(1,-1,1).$$

Its dot product with each edge is zero, so it is perpendicular to the triangular face. Its length is $\sqrt3$, the area of the parallelogram spanned by the edges. The triangle has half that area, $\sqrt3/2$, and dividing the normal by $\sqrt3$ makes it a unit normal.

Order carries orientation: $v\times u=-(u\times v)$. If the edges become parallel, the triangle collapses and the cross product is zero. There is then no unique unit normal to recover by dividing by its length. The code reports that case explicitly.

```python
u = np.array([1,1,0]); v = np.array([0,1,1])
for first, second in [(u,v),(v,u),(u,2*u)]:
    normal, unit, area = la.normal_and_area(first,second)
    print('normal:',normal,'unit normal:',unit,'triangle area:',area)
normal, unit, area = la.normal_and_area(u,v)
fig = plt.figure(figsize=(6,5)); ax3 = fig.add_subplot(projection='3d')
for vec,label in [(u,'u'),(v,'v'),(normal,'u × v')]:
    ax3.quiver(0,0,0,*vec,label=label)
ax3.set(xlim=(-2,2),ylim=(-2,2),zlim=(-2,2),xlabel='x',ylabel='y',zlabel='z')
ax3.set_box_aspect((1,1,1)); ax3.legend(); display(fig); plt.close(fig)
```
