## Square symmetry challenge

Before running this check, list the eight matrices by hand. Guess whether rotations and reflections commute. Use a basis vector to disprove your guess if necessary. The code verifies this finite set; your seminar proof explains why the reduction rule always works.

### Why eight transformations form one group

$R$ is a quarter-turn and $F$ reflects in the x-axis. Four quarter-turns return every vector to itself, so $R^4=I$. Reflecting twice also returns it, so $F^2=I$.

Now follow one point through “rotate, then reflect.” The reflection reverses the sense of rotation, giving $FR=R^{-1}F$. This relation lets us move every reflection to the right of a product. Any composition reduces to $R^k$ or $R^kF$, with $k=0,1,2,3$. For example,

$$ (R^iF)(R^jF)=R^iR^{-j}F^2=R^{i-j}.$$

This is a structural explanation of closure. The following code checks all products for this small finite set; it does not replace the explanation. To disprove commutativity, one vector is enough: $RF e_1=e_2$ but $FR e_1=-e_2$.

```python
R = np.array([[0,-1],[1,0]], dtype=int)
F = np.diag([1,-1])
rotations = [np.linalg.matrix_power(R,k) for k in range(4)]
group = rotations + [M@F for M in rotations]
assert len({tuple(M.ravel()) for M in group}) == 8
assert all(any(np.array_equal(A@B,C) for C in group) for A in group for B in group)
assert all(any(np.array_equal(A@B,np.eye(2)) for B in group) for A in group)
print('RF =\n',R@F,'\nFR =\n',F@R)
print('All 64 products belong to the set.')
```
